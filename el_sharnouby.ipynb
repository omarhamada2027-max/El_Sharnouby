```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
"D:\visual projects\Sharnouby_data"
```

```python

data_dir = r"D:\visual projects\Sharnouby_data" 

dfs = {}

for file in os.listdir(data_dir):
    if file.endswith('.csv'):
        file_path = os.path.join(data_dir, file)
        # استخراج اسم الجدول بدون .csv لاستخدامه كمعرف
        table_name = file.replace('.csv', '')
        try:
            dfs[table_name] = pd.read_csv(file_path)
            print(f"Loaded successfully: {file} | Shape: {dfs[table_name].shape}")
        except Exception as e:
            print(f"Error loading {file}: {e}")

print(f"\nTotal tables loaded: {len(dfs)}")
```

```python

business_units_df = dfs['Business_Units']
case_briefs_df = dfs['Case_Briefs']
customer_service_df = dfs['Customer_Service']
customers_df = dfs['Customers']
data_dictionary_df = dfs['Data_Dictionary']
group_operations_df = dfs['Group_Operations']
inventory_history_df = dfs['Inventory_History']
locations_df = dfs['Locations']
machine_downtime_df = dfs['Machine_Downtime']
marketing_df = dfs['Marketing']
materials_df = dfs['Materials']
printing_batches_df = dfs['Printing_Batches']
product_bom_df = dfs['Product_BOM']
product_master_df = dfs['Product_Master']
production_orders_df = dfs['Production_Orders']
purchase_orders_df = dfs['Purchase_Orders']
quality_df = dfs['Quality']
sales_transactions_df = dfs['Sales_Transactions']
suppliers_df = dfs['Suppliers']
web_traffic_daily_df = dfs['Web_Traffic_Daily']
```

```python

found_nulls = False

for table_name, df in dfs.items():
    null_counts = df.isnull().sum()
    null_counts = null_counts[null_counts > 0] # الأعمدة التي بها نالس فقط
    
    if not null_counts.empty:
        found_nulls = True
        print(f" {table_name}")
        for col, count in null_counts.items():
            pct = (count / len(df)) * 100
            print(f"{col}': {count} قيمة مفقودة ({pct:.2f}%)")

if not found_nulls:
    print("\n مبروك! لا توجد أي قيم مفقودة في أي جدول بالكامل.")
```

```python

total_net_sales = sales_transactions_df['Net_Sales_EGP'].sum()
total_cogs = sales_transactions_df['Estimated_COGS_EGP'].sum()
total_gross_profit = sales_transactions_df['Gross_Profit_EGP'].sum()
overall_margin = (total_gross_profit / total_net_sales) * 100

print("--- Executive Financial KPIs ---")
print(f"Total Net Sales: {total_net_sales:,.2f} EGP")
print(f"Total COGS: {total_cogs:,.2f} EGP")
print(f"Total Gross Profit: {total_gross_profit:,.2f} EGP")
print(f"Overall Gross Margin: {overall_margin:.2f}%")

# 2. Channel Performance Analysis
channel_performance = sales_transactions_df.groupby('Channel').agg(
    Net_Sales=('Net_Sales_EGP', 'sum'),
    Gross_Profit=('Gross_Profit_EGP', 'sum'),
    Transactions_Count=('Invoice_ID', 'count')
).reset_index()

channel_performance['Margin_Pct'] = (channel_performance['Gross_Profit'] / channel_performance['Net_Sales']) * 100

print("\n--- Performance by Sales Channel ---")
display(channel_performance)
```

```python
monthly_performance = sales_transactions_df.groupby('Month_Key').agg(
    Net_Sales=('Net_Sales_EGP', 'sum'),
    Gross_Profit=('Gross_Profit_EGP', 'sum'),
    Quantity=('Quantity', 'sum')
).reset_index().sort_values('Month_Key')

print("--- Monthly Sales & Profitability Trend ---")
display(monthly_performance)
```

```python
top_customers = sales_transactions_df.groupby(['Customer_ID', 'Customer_Type', 'Customer_Governorate']).agg(
    Total_Spent=('Net_Sales_EGP', 'sum'),
    Total_Orders=('Invoice_ID', 'count')
).reset_index().sort_values(by='Total_Spent', ascending=False).head(5)

print("--- Top 5 Customers by Revenue ---")
display(top_customers)
```

```python
governorate_performance = sales_transactions_df.groupby('Customer_Governorate').agg(
    Net_Sales=('Net_Sales_EGP', 'sum'),
    Gross_Profit=('Gross_Profit_EGP', 'sum')
).reset_index().sort_values(by='Net_Sales', ascending=False)

print("--- Sales & Profit by Governorate ---")
display(governorate_performance)
```

```python
returns_summary = sales_transactions_df.groupby('Returned').agg(
    Transaction_Count=('Invoice_ID', 'count'),
    Total_Refund_Value=('Refund_Value_EGP', 'sum')
).reset_index()

print("--- Returns Impact Summary ---")
display(returns_summary)
```

```python
# Production Orders Bottleneck Analysis (Corrected Column Names)
bottleneck_analysis = production_orders_df.groupby('Primary_Bottleneck').agg(
    Order_Count=('Production_Order_ID', 'count'),
    Total_Quantity=('Planned_Qty', 'sum')
).reset_index().sort_values(by='Order_Count', ascending=False)

print("--- Production Bottlenecks Analysis ---")
display(bottleneck_analysis)
```

```python
# Printing Performance by Print Type
printing_performance = printing_batches_df.groupby('Print_Type').agg(
    Total_Batches=('Print_Batch_ID', 'count'),
    Planned_Quantity=('Planned_Qty', 'sum'),
    Printed_Quantity=('Printed_Qty', 'sum'),
    Total_Ink_Cost=('Ink_or_Consumable_Cost_EGP', 'sum'),
    Total_Rework_Cost=('Rework_Cost_EGP', 'sum')
).reset_index()

print("--- Printing Performance by Type ---")
display(printing_performance)
```

```python
# Inventory Valuation & Stock Levels by Location
inventory_summary = inventory_history_df.groupby(['Location_ID', 'Location_Name']).agg(
    Total_SKUs=('SKU', 'count'),
    Total_Closing_Stock=('Closing_Stock', 'sum'),
    Total_Stockout_Days=('Stockout_Days_In_Month', 'sum')
).reset_index().sort_values(by='Total_Closing_Stock', ascending=False)

print("--- Inventory Levels and Stockout Summary by Location ---")
display(inventory_summary)
```

```python
# Customer Segmentation Analysis by Type, Governorate, and Channel
customer_segmentation = customers_df.groupby(['Customer_Type', 'Governorate', 'Acquisition_Channel']).agg(
    Total_Customers=('Customer_ID', 'count')
).reset_index().sort_values(by='Total_Customers', ascending=False)

print("--- Customer Segmentation Summary ---")
display(customer_segmentation)
```

```python
# Marketing Campaigns Performance Analysis by Platform
marketing_df = dfs['Marketing']

marketing_performance = marketing_df.groupby('Platform').agg(
    Total_Campaigns=('Campaign_ID', 'count'),
    Total_Spend=('Spend_EGP', 'sum'),
    Total_Revenue=('Attributed_Revenue_EGP', 'sum'),
    Total_Clicks=('Clicks', 'sum'),
    Total_Impressions=('Impressions', 'sum')
).reset_index()

# Calculate Return on Ad Spend (ROAS)
marketing_performance['ROAS'] = marketing_performance['Total_Revenue'] / marketing_performance['Total_Spend']

print("--- Marketing Performance Summary by Platform ---")
display(marketing_performance)
```

```python
# Customer Service Performance Analysis (Corrected Column Names)
cs_df = dfs['Customer_Service']

cs_summary = cs_df.groupby(['Issue_Type', 'Resolved']).agg(
    Total_Tickets=('Ticket_ID', 'count'),
    Avg_Resolution_Hours=('Resolution_Hours', 'mean'),
    Avg_CSAT=('CSAT_1_to_5', 'mean')
).reset_index().sort_values(by='Total_Tickets', ascending=False)

print("--- Customer Service Summary ---")
display(cs_summary)
```

```python
# Quality Inspections and Defect Loss Analysis (Corrected Column Names)
quality_df = dfs['Quality']

quality_summary = quality_df.groupby(['Stage', 'Defect_Type']).agg(
    Total_Inspections=('QC_ID', 'count'),
    Total_Inspected_Qty=('Inspected_Qty', 'sum'),
    Total_Rejected_Qty=('Rejected_Qty', 'sum'),
    Total_Estimated_Loss=('Estimated_Loss_EGP', 'sum')
).reset_index().sort_values(by='Total_Rejected_Qty', ascending=False)

print("--- Quality Defect and Loss Summary ---")
display(quality_summary)
```

```python
# Purchase Orders and Supplier Performance Analysis (Corrected Column Names)
po_df = dfs['Purchase_Orders']

# Calculate Total Cost per row
po_df['Total_Cost_EGP'] = po_df['Ordered_Qty'] * po_df['Unit_Cost_EGP']

supplier_summary = po_df.groupby('Supplier_ID').agg(
    Total_POs=('PO_ID', 'count'),
    Total_Ordered_Qty=('Ordered_Qty', 'sum'),
    Total_Accepted_Qty=('Accepted_Qty', 'sum'),
    Total_Rejected_Qty=('Rejected_Qty', 'sum'),
    Total_Spend=('Total_Cost_EGP', 'sum')
).reset_index().sort_values(by='Total_Spend', ascending=False)

print("--- Purchase Orders & Supplier Summary ---")
display(supplier_summary)
```

```python
# Daily Web Traffic and Conversion Analysis (Corrected)
web_df = dfs['Web_Traffic_Daily']

web_summary = pd.DataFrame({
    'Metric': ['Total Sessions', 'Total Users', 'Total Orders', 'Avg Conversion Rate', 'Avg Add To Cart Rate', 'Avg Page Load Sec'],
    'Value': [
        web_df['Sessions'].sum(),
        web_df['Users'].sum(),
        web_df['Orders'].sum(),
        web_df['Conversion_Rate'].mean(),
        web_df['Add_To_Cart_Rate'].mean(),
        web_df['Avg_Page_Load_Sec'].mean()
    ]
})

print("--- Web Traffic Daily Overall Performance Summary ---")
display(web_summary)
```
